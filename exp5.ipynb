```python

# ================================

# Install (usually preinstalled in Colab, but safe)
!pip install -q tensorflow matplotlib pandas

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
import pandas as pd

print("TensorFlow version:", tf.__version__)

# ================================
# 1. LOAD DATA
# ================================
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

# ================================
# 2. MODEL DEFINITIONS
# ================================
def build_model(model_type="dense"):

    if model_type == "dense":
        model = keras.Sequential([
            layers.Flatten(input_shape=(28, 28)),
            layers.Dense(128, activation='relu'),
            layers.Dense(10, activation='softmax')
        ])

    elif model_type == "deep":
        model = keras.Sequential([
            layers.Flatten(input_shape=(28, 28)),
            layers.Dense(256, activation='relu'),
            layers.Dense(128, activation='relu'),
            layers.Dense(64, activation='relu'),
            layers.Dense(10, activation='softmax')
        ])

    elif model_type == "cnn":
        model = keras.Sequential([
            layers.Reshape((28, 28, 1), input_shape=(28, 28)),
            layers.Conv2D(32, (3,3), activation='relu'),
            layers.MaxPooling2D(),
            layers.Conv2D(64, (3,3), activation='relu'),
            layers.MaxPooling2D(),
            layers.Flatten(),
            layers.Dense(64, activation='relu'),
            layers.Dense(10, activation='softmax')
        ])

    return model

# ================================
# 3. TRAIN FUNCTION
# ================================
def train_and_evaluate(model_type, optimizer_name, epochs=5):

    optimizer_dict = {
        "adam": keras.optimizers.Adam(),
        "sgd": keras.optimizers.SGD(),
        "rmsprop": keras.optimizers.RMSprop()
    }

    model = build_model(model_type)

    model.compile(
        optimizer=optimizer_dict[optimizer_name],
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    history = model.fit(
        x_train, y_train,
        epochs=epochs,
        validation_split=0.1,
        verbose=1
    )

    test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)

    return history, test_acc

# ================================
# 4. RUN EXPERIMENTS
# ================================
experiments = []
histories = {}

model_types = ["dense", "deep", "cnn"]
optimizers = ["adam", "sgd", "rmsprop"]

for m in model_types:
    for opt in optimizers:
        print(f"\nRunning: {m} + {opt}")

        history, acc = train_and_evaluate(m, opt)

        key = f"{m}_{opt}"
        histories[key] = history

        experiments.append({
            "model": m,
            "optimizer": opt,
            "test_accuracy": acc
        })

# ================================
# 5. RESULTS TABLE
# ================================
df = pd.DataFrame(experiments)
df = df.sort_values(by="test_accuracy", ascending=False)

print("\n=== FINAL RESULTS ===")
print(df)

# ================================
# 6. PLOT FUNCTION
# ================================
def plot_history(history, title):
    plt.figure()
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title(title)
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.show()

# ================================
# 7. PLOT SOME KEY MODELS
# ================================
plot_history(histories["cnn_adam"], "CNN + Adam")
plot_history(histories["dense_sgd"], "Dense + SGD")
plot_history(histories["deep_rmsprop"], "Deep + RMSprop")

# ================================
# 8. TRAIN & SAVE BEST MODEL (CNN + Adam)
# ================================
best_model = build_model("cnn")

best_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

best_model.fit(x_train, y_train, epochs=5, verbose=1)

best_model.save("best_mnist_model.h5")

print("\nBest model saved as 'best_mnist_model.h5'")
```
